### 1.1 Baseline- Training ResNet

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import wandb
import matplotlib.pyplot as plt
import numpy as np
from torch.utils.data import DataLoader, TensorDataset
import seaborn as sns
from sklearn.metrics import f1_score, confusion_matrix
# Initialize Wandb for experiment logging


# Load the dataset
train_data = torch.load('Dataset/Q1/train_data.pt')
train_labels = torch.load('Dataset/Q1/train_labels.pt')
test_data = torch.load('Dataset/Q1/test_data.pt')
test_labels = torch.load('Dataset/Q1/test_labels.pt')

# Normalize the data for training
transform = transforms.Compose([transforms.ToTensor(), transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])])
train_data = train_data.float() / 255.0  # Scale to [0, 1] and convert to float
test_data = test_data.float() / 255.0  # Scale to [0, 1] and convert to float

# Create dataset and dataloaders
train_dataset = TensorDataset(train_data, train_labels)
test_dataset = TensorDataset(test_data, test_labels)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Load ResNet18 model
resnet18 = torchvision.models.resnet18(pretrained=False)  # Standard ResNet18

# Modify the final layer to match the number of classes in the dataset
num_classes = len(torch.unique(train_labels))  # Example, you should set this based on your dataset
resnet18.fc = nn.Linear(resnet18.fc.in_features, num_classes)

# Loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(resnet18.parameters(), lr=0.00001)

# Device configuration (use GPU if available)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
resnet18 = resnet18.to(device)

# Function to train the model
def train_model(model, train_loader, test_loader, criterion, optimizer, num_epochs=10,flag=0):
    wandb.init(project=f"ResNet18-Experiment_1.1_flag={flag}")
    model.train()
    all_train_losses = []
    all_train_accuracies = []
    all_test_losses = []
    all_test_accuracies = []
    
    for epoch in range(num_epochs):
        running_loss = 0.0
        correct = 0
        total = 0
        
        # Training phase
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            
            optimizer.zero_grad()
            
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
        
        train_loss = running_loss / len(train_loader)
        train_accuracy = 100 * correct / total
        
        # Test phase
        model.eval()
        all_preds = []
        all_labels = []
        with torch.no_grad():
            correct = 0
            total = 0
            test_loss = 0.0
            for inputs, labels in test_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                test_loss += loss.item()
                
                _, predicted = torch.max(outputs, 1)
                total += labels.size(0)
                correct += (predicted == labels).sum().item()

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())
        
        test_loss /= len(test_loader)
        test_accuracy = 100 * correct / total

        if epoch == num_epochs - 1:
            cm = confusion_matrix(all_labels, all_preds)
            plt.figure(figsize=(10, 8))
            sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                       cbar=True, square=True, 
                       xticklabels=range(10), 
                       yticklabels=range(10))
            plt.title(f'Confusion Matrix - ResNet18-Experiment_1.1')
            plt.xlabel('Predicted Label')
            plt.ylabel('True Label')
            filename = f'confusion_matrix_ResNet18-Experiment_1.1_flag{flag}.png'
            plt.savefig(filename, bbox_inches='tight', dpi=300)
            plt.close()
            
            print(f"Saved confusion matrix to {filename}")
        
        # Logging to Wandb
        wandb.log({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "train_accuracy": train_accuracy,
            "test_loss": test_loss,
            "test_accuracy": test_accuracy
        })
        
        all_train_losses.append(train_loss)
        all_train_accuracies.append(train_accuracy)
        all_test_losses.append(test_loss)
        all_test_accuracies.append(test_accuracy)
        
        print(f"Epoch {epoch + 1}/{num_epochs} - "
              f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_accuracy:.2f}%, "
              f"Test Loss: {test_loss:.4f}, Test Accuracy: {test_accuracy:.2f}%")
    # Final log to Wandb

    wandb.finish()
    return all_train_losses, all_train_accuracies, all_test_losses, all_test_accuracies

# Train the model
train_losses, train_accuracies, test_losses, test_accuracies = train_model(
    resnet18, train_loader, test_loader, criterion, optimizer, num_epochs=15,flag="0"
)

# Now, retrain with a pretrained ResNet18 (on ImageNet) and modify the last layer
pretrained_resnet18 = torchvision.models.resnet18(pretrained=True)
pretrained_resnet18.fc = nn.Linear(pretrained_resnet18.fc.in_features, num_classes)
pretrained_resnet18 = pretrained_resnet18.to(device)

# Train the pretrained model
optimizer_pretrained = optim.Adam(pretrained_resnet18.parameters(), lr=1e-5)
train_losses_pretrained, train_accuracies_pretrained, test_losses_pretrained, test_accuracies_pretrained = train_model(
    pretrained_resnet18, train_loader, test_loader, criterion, optimizer_pretrained, num_epochs=15,flag="1"
)

# Plot accuracy and loss for both models
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(range(1, 16), train_accuracies, label="Train Accuracy (Standard)")
plt.plot(range(1, 16), test_accuracies, label="Test Accuracy (Standard)")
plt.plot(range(1, 16), train_accuracies_pretrained, label="Train Accuracy (Pretrained)")
plt.plot(range(1, 16), test_accuracies_pretrained, label="Test Accuracy (Pretrained)")
plt.title("Accuracy Over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(range(1, 16), train_losses, label="Train Loss (Standard)")
plt.plot(range(1, 16), test_losses, label="Test Loss (Standard)")
plt.plot(range(1, 16), train_losses_pretrained, label="Train Loss (Pretrained)")
plt.plot(range(1, 16), test_losses_pretrained, label="Test Loss (Pretrained)")
plt.title("Loss Over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.tight_layout()
plt.show()


print("final_train_accuracy_scratch", train_accuracies[-1],
    "\nfinal_test_accuracy_scratch", test_accuracies[-1],
    "\nfinal_train_accuracy_pretrained", train_accuracies_pretrained[-1],
    "\nfinal_test_accuracy_pretrained", test_accuracies_pretrained[-1])
wandb.finish()
```

### 1.2 Training ResNet on resized images

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import wandb
import matplotlib.pyplot as plt
import numpy as np
from torch.utils.data import DataLoader, TensorDataset
import seaborn as sns
from sklearn.metrics import f1_score, confusion_matrix
# Initialize Wandb for experiment logging


# Load the dataset
train_data = torch.load('Dataset/Q1/train_data.pt')
train_labels = torch.load('Dataset/Q1/train_labels.pt')
test_data = torch.load('Dataset/Q1/test_data.pt')
test_labels = torch.load('Dataset/Q1/test_labels.pt')

# Resize the images to 224x224 and apply normalization
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])  # ImageNet stats
])


# Apply transformations to train and test data
train_data_resized = []
for image in train_data:
    pil_image = transforms.ToPILImage()(image)  # Convert tensor to PIL Image
    train_data_resized.append(transform(pil_image))  # Apply transformations

test_data_resized = []
for image in test_data:
    pil_image = transforms.ToPILImage()(image)  # Convert tensor to PIL Image
    test_data_resized.append(transform(pil_image))  # Apply transformations

# Convert the list of transformed images back into a single tensor
train_data_resized = torch.stack(train_data_resized)
test_data_resized = torch.stack(test_data_resized)

# Create dataset and dataloaders
train_dataset = TensorDataset(train_data_resized, train_labels)
test_dataset = TensorDataset(test_data_resized, test_labels)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Load ResNet18 model (from scratch)
resnet18 = torchvision.models.resnet18(pretrained=False)  # Standard ResNet18

# Modify the final layer to match the number of classes in the dataset
num_classes = len(torch.unique(train_labels))  # Example, set based on your dataset
resnet18.fc = nn.Linear(resnet18.fc.in_features, num_classes)

# Loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(resnet18.parameters(), lr=0.00001)

# Device configuration (use GPU if available)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
resnet18 = resnet18.to(device)

# Function to train the model
def train_model(model, train_loader, test_loader, criterion, optimizer, num_epochs=10,flag="0"):
    wandb.init(project=f"ResNet18-Resized-Experiment_1.2_flag={flag}")
    model.train()
    all_train_losses = []
    all_train_accuracies = []
    all_test_losses = []
    all_test_accuracies = []
    
    for epoch in range(num_epochs):
        running_loss = 0.0
        correct = 0
        total = 0
        
        # Training phase
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            
            optimizer.zero_grad()
            
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            running_loss += loss.item()
            
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
        
        train_loss = running_loss / len(train_loader)
        train_accuracy = 100 * correct / total
        
        # Test phase
        model.eval()
        all_preds = []
        all_labels = []
        with torch.no_grad():
            correct = 0
            total = 0
            test_loss = 0.0
            for inputs, labels in test_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                test_loss += loss.item()
                
                _, predicted = torch.max(outputs, 1)
                total += labels.size(0)
                correct += (predicted == labels).sum().item()

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())
        
        test_loss /= len(test_loader)
        test_accuracy = 100 * correct / total

        if epoch == num_epochs - 1:
            cm = confusion_matrix(all_labels, all_preds)
            plt.figure(figsize=(10, 8))
            sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                       cbar=True, square=True, 
                       xticklabels=range(10), 
                       yticklabels=range(10))
            plt.title(f'Confusion Matrix - ResNet18-Resized-Experiment_1.2')
            plt.xlabel('Predicted Label')
            plt.ylabel('True Label')
            filename = f'confusion_matrix_ResNet18-Resized-Experiment_1.2_flag{flag}.png'
            plt.savefig(filename, bbox_inches='tight', dpi=300)
            plt.close()
            
            print(f"Saved confusion matrix to {filename}")
        
        # Logging to Wandb
        wandb.log({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "train_accuracy": train_accuracy,
            "test_loss": test_loss,
            "test_accuracy": test_accuracy
        })
        
        all_train_losses.append(train_loss)
        all_train_accuracies.append(train_accuracy)
        all_test_losses.append(test_loss)
        all_test_accuracies.append(test_accuracy)
        
        print(f"Epoch {epoch + 1}/{num_epochs} - "
              f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_accuracy:.2f}%, "
              f"Test Loss: {test_loss:.4f}, Test Accuracy: {test_accuracy:.2f}%")
    wandb.finish()
    return all_train_losses, all_train_accuracies, all_test_losses, all_test_accuracies

# Train the ResNet18 from scratch on resized images
train_losses_resnet, train_accuracies_resnet, test_losses_resnet, test_accuracies_resnet = train_model(
    resnet18, train_loader, test_loader, criterion, optimizer, num_epochs=15,flag="0"
)

# Now, retrain with a pretrained ResNet18 (on ImageNet) and modify the last layer
pretrained_resnet18 = torchvision.models.resnet18(pretrained=True)
pretrained_resnet18.fc = nn.Linear(pretrained_resnet18.fc.in_features, num_classes)
pretrained_resnet18 = pretrained_resnet18.to(device)

# Train the pretrained model on resized images
optimizer_pretrained = optim.Adam(pretrained_resnet18.parameters(), lr=1e-5)
train_losses_pretrained, train_accuracies_pretrained, test_losses_pretrained, test_accuracies_pretrained = train_model(
    pretrained_resnet18, train_loader, test_loader, criterion, optimizer_pretrained, num_epochs=15,flag="1"
)

# Plot accuracy and loss for both models
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(range(1, 16), train_accuracies_resnet, label="Train Accuracy (Scratch)")
plt.plot(range(1, 16), test_accuracies_resnet, label="Test Accuracy (Scratch)")
plt.plot(range(1, 16), train_accuracies_pretrained, label="Train Accuracy (Pretrained)")
plt.plot(range(1, 16), test_accuracies_pretrained, label="Test Accuracy (Pretrained)")
plt.title("Accuracy Over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(range(1, 16), train_losses_resnet, label="Train Loss (Scratch)")
plt.plot(range(1, 16), test_losses_resnet, label="Test Loss (Scratch)")
plt.plot(range(1, 16), train_losses_pretrained, label="Train Loss (Pretrained)")
plt.plot(range(1, 16), test_losses_pretrained, label="Test Loss (Pretrained)")
plt.title("Loss Over Epochs")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.tight_layout()
plt.show()

# Final log to Wandb

print("final_train_accuracy_scratch", train_accuracies_resnet[-1],
    "\nfinal_test_accuracy_scratch", test_accuracies_resnet[-1],
    "\nfinal_train_accuracy_pretrained", train_accuracies_pretrained[-1],
    "\nfinal_test_accuracy_pretrained", test_accuracies_pretrained[-1])
wandb.finish()
```

### 1.3 Modifying the architecture of ResNet18 to suit the given dataset 

ResNet18Mod1 (Trained from scratch):
- Single 3x3 conv (3→64 channels) replacing 7x7 conv
- Stride=1, padding=1, bias=False
- Removes maxpool entirely (Identity)
- Purpose: Preserves spatial dimensions for small inputs

ResNet18Mod2 (Trained from scratch):
- Two sequential 3x3 conv layers
- First layer: 3→32 channels with BatchNorm + ReLU
- Second layer: 32→64 channels
- Replaces maxpool with 2x2 stride-2 pool
- Purpose: Gradual feature extraction with controlled downsampling

ResNet18Mod3 (Trained from scratch):
- Triple 3x3 conv sequence
- Channel progression: 3→64→64→64
- Each conv followed by BatchNorm (last conv's BatchNorm without ReLU)
- 2x2 stride-2 maxpool
- Purpose: Deep initial feature extraction with normalization

PretrainedMod1:
- Loads pretrained weights except first layer
- New 3x3 conv (3→64) replacing original 7x7
- 2x2 stride-2 maxpool
- Purpose: Adapts input layer while preserving pretrained features

PretrainedMod2:
- Custom input pathway: two 3x3 convs (3→32→64)
- First conv has BatchNorm + ReLU
- Disables original conv1 and maxpool (Identity)
- Purpose: Domain-specific input processing before pretrained layers

PretrainedMod3:
- Parallel processing paths:
  * Path1: 3x3 conv (3→32) + BatchNorm + ReLU
  * Path2: 5x5 conv (3→32) + BatchNorm + ReLU
- Concatenates paths (64 channels)
- 1x1 conv fusion layer (64→64)
- Disables original conv1 and maxpool
- Purpose: Multi-scale feature extraction before pretrained layers

```python
import torch
import torch.nn as nn
import torchvision.models as models
from torch.utils.data import DataLoader, TensorDataset
import matplotlib.pyplot as plt
from sklearn.metrics import f1_score, confusion_matrix
import wandb
import seaborn as sns
# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
# Load the dataset
train_data = torch.load('Dataset/Q1/train_data.pt')
train_labels = torch.load('Dataset/Q1/train_labels.pt')
test_data = torch.load('Dataset/Q1/test_data.pt')
test_labels = torch.load('Dataset/Q1/test_labels.pt')
# Data Loading and Preprocessing
def prepare_data(train_data, train_labels, test_data, test_labels, batch_size=32):
    train_dataset = TensorDataset(train_data, train_labels)
    test_dataset = TensorDataset(test_data, test_labels)
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
    return train_loader, test_loader

# Part 1: Models trained from scratch
class ResNet18Mod1(nn.Module):
    """Modification 1: Smaller kernel and stride, no maxpool"""
    def __init__(self, num_classes=10):
        super(ResNet18Mod1, self).__init__()
        self.model = models.resnet18(pretrained=False)
        # Modify first conv layer to use 3x3 kernel with stride 1
        self.model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
        # Remove maxpool
        self.model.maxpool = nn.Identity()
        # Adjust final layer
        self.model.fc = nn.Linear(512, num_classes)
    
    def forward(self, x):
        return self.model(x)

class ResNet18Mod2(nn.Module):
    """Modification 2: Double conv with smaller kernels"""
    def __init__(self, num_classes=10):
        super(ResNet18Mod2, self).__init__()
        self.model = models.resnet18(pretrained=False)
        # Replace conv1 with two smaller conv layers
        self.model.conv1 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1)
        )
        # Replace maxpool with smaller pool
        self.model.maxpool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.model.fc = nn.Linear(512, num_classes)
    
    def forward(self, x):
        return self.model(x)

class ResNet18Mod3(nn.Module):
    """Modification 3: Triple conv with residual connection"""
    def __init__(self, num_classes=10):
        super(ResNet18Mod3, self).__init__()
        self.model = models.resnet18(pretrained=False)
        # Custom initial block
        self.model.conv1 = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Conv2d(64, 64, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(64)
        )
        # Smaller maxpool
        self.model.maxpool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.model.fc = nn.Linear(512, num_classes)
    
    def forward(self, x):
        return self.model(x)

# Part 2: Pretrained model modifications
class PretrainedMod1(nn.Module):
    """Pretrained Modification 1: Adapted first layer with dimensionality preservation"""
    def __init__(self, num_classes=10):
        super(PretrainedMod1, self).__init__()
        self.model = models.resnet18(pretrained=True)
        # New first conv layer (will be initialized from scratch)
        self.model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1)
        # Modified pooling
        self.model.maxpool = nn.MaxPool2d(kernel_size=2, stride=2)
        self.model.fc = nn.Linear(512, num_classes)
    
    def forward(self, x):
        return self.model(x)

class PretrainedMod2(nn.Module):
    """Pretrained Modification 2: Custom input pathway"""
    def __init__(self, num_classes=10):
        super(PretrainedMod2, self).__init__()
        self.model = models.resnet18(pretrained=True)
        # Custom input pathway
        self.input_pathway = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1)
        )
        # Disable original first conv and pool
        self.model.conv1 = nn.Identity()
        self.model.maxpool = nn.Identity()
        self.model.fc = nn.Linear(512, num_classes)
    
    def forward(self, x):
        x = self.input_pathway(x)
        return self.model(x)

class PretrainedMod3(nn.Module):
    """Pretrained Modification 3: Parallel pathways with fusion"""
    def __init__(self, num_classes=10):
        super(PretrainedMod3, self).__init__()
        self.model = models.resnet18(pretrained=True)
        
        # Two parallel pathways
        self.path1 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, stride=1, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        self.path2 = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=5, stride=1, padding=2),
            nn.BatchNorm2d(32),
            nn.ReLU()
        )
        # Fusion layer
        self.fusion = nn.Conv2d(64, 64, kernel_size=1)
        
        # Disable original first layers
        self.model.conv1 = nn.Identity()
        self.model.maxpool = nn.Identity()
        self.model.fc = nn.Linear(512, num_classes)
    
    def forward(self, x):
        # Parallel processing
        p1 = self.path1(x)
        p2 = self.path2(x)
        # Concatenate and fuse
        x = torch.cat([p1, p2], dim=1)
        x = self.fusion(x)
        return self.model(x)

# Training function with metrics
def train_and_evaluate(model, train_loader, test_loader, criterion, optimizer, num_epochs=10,project_name="ResNet18-Experiment"):
    model = model.to(device)
    train_losses, train_accs = [], []
    test_losses, test_accs = [], []
    best_acc = 0

    wandb.init(project=project_name, config={"epochs": num_epochs})
    
    for epoch in range(num_epochs):
        # Training
        model.train()
        train_loss = 0
        correct = 0
        total = 0
        
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            inputs = inputs.float()
            labels =labels
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
            
            train_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
        
        train_loss = train_loss / len(train_loader)
        train_acc = 100. * correct / total
        train_losses.append(train_loss)
        train_accs.append(train_acc)
        
        # Testing
        model.eval()
        test_loss = 0
        correct = 0
        total = 0
        all_preds = []
        all_labels = []
        
        with torch.no_grad():
            for inputs, labels in test_loader:
                inputs, labels = inputs.to(device), labels.to(device)
                inputs = inputs.float()
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                
                test_loss += loss.item()
                _, predicted = outputs.max(1)
                total += labels.size(0)
                correct += predicted.eq(labels).sum().item()

                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())
        
        test_loss = test_loss / len(test_loader)
        test_acc = 100. * correct / total
        test_losses.append(test_loss)
        test_accs.append(test_acc)
        
        # Calculate F1 score and confusion matrix
        f1 = f1_score(all_labels, all_preds, average='weighted')
        
        if epoch == num_epochs - 1:
            cm = confusion_matrix(all_labels, all_preds)
            plt.figure(figsize=(10, 8))
            sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                       cbar=True, square=True, 
                       xticklabels=range(10), 
                       yticklabels=range(10))
            plt.title(f'Confusion Matrix - {project_name}')
            plt.xlabel('Predicted Label')
            plt.ylabel('True Label')
            filename = f'confusion_matrix_{project_name}.png'
            plt.savefig(filename, bbox_inches='tight', dpi=300)
            plt.close()
            
            print(f"Saved confusion matrix to {filename}")

        
        # Log metrics to WandB
        wandb.log({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "train_accuracy": train_acc,
            "test_loss": test_loss,
            "test_accuracy": test_acc,
            "f1_score": f1,
        })

        print(f'Epoch {epoch}/{num_epochs}: Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}%, '
              f'Test Loss: {test_loss:.4f}, Test Acc: {test_acc:.2f}%, F1 Score: {f1:.4f}')

        
        # Save best model
        if test_acc > best_acc:
            best_acc = test_acc
            torch.save(model.state_dict(), f'best_{model.__class__.__name__}.pth')
    wandb.finish()
    return train_losses, train_accs, test_losses, test_accs

# Plotting function
def plot_training_curves(metrics_dict):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    for name, metrics in metrics_dict.items():
        train_losses, train_accs, test_losses, test_accs = metrics
        ax1.plot(train_losses, label=f'{name} (train)')
        ax1.plot(test_losses, label=f'{name} (test)')
    ax1.set_title('Loss Curves')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.legend()
    
    for name, metrics in metrics_dict.items():
        train_losses, train_accs, test_losses, test_accs = metrics
        ax2.plot(train_accs, label=f'{name} (train)')
        ax2.plot(test_accs, label=f'{name} (test)')
    ax2.set_title('Accuracy Curves')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Accuracy (%)')
    ax2.legend()
    
    plt.tight_layout()
    plt.show()

# Main execution
def main():
    # Prepare data
    train_loader, test_loader = prepare_data(train_data, train_labels, test_data, test_labels, batch_size=256)
    
    # Training parameters
    num_epochs = 10
    criterion = nn.CrossEntropyLoss()
    
    # Dictionary to store all metrics
    all_metrics = {}
    
    # Train models from scratch
    scratch_models = [
        ('Scratch_Mod1', ResNet18Mod1(), "ResNet18_Scratch_Mod1"),
        ('Scratch_Mod2', ResNet18Mod2(), "ResNet18_Scratch_Mod2"),
        ('Scratch_Mod3', ResNet18Mod3(), "ResNet18_Scratch_Mod3")
    ]
    
    for name, model, project_name in scratch_models:
        print(f"\nTraining {name}")
        optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
        metrics = train_and_evaluate(model, train_loader, test_loader, criterion, optimizer, num_epochs=num_epochs, project_name=project_name)
        all_metrics[name] = metrics
    
    # Train pretrained models
    pretrained_models = [
        ('Pretrained_Mod1', PretrainedMod1(), "ResNet18_Pretrained_Mod1"),
        ('Pretrained_Mod2', PretrainedMod2(), "ResNet18_Pretrained_Mod2"),
        ('Pretrained_Mod3', PretrainedMod3(), "ResNet18_Pretrained_Mod3")
    ]
    
    for name, model, project_name in pretrained_models:
        print(f"\nTraining {name}")
        # Different learning rates for different parts
        base_params = list(model.parameters())[:-2]  # excluding final layers
        final_params = list(model.parameters())[-2:]  # final layers
        optimizer = torch.optim.Adam([ 
            {'params': base_params, 'lr': 0.0001},  # lower learning rate for pretrained parts
            {'params': final_params, 'lr': 0.001}   # higher learning rate for new parts
        ])
        metrics = train_and_evaluate(model, train_loader, test_loader, criterion, optimizer, num_epochs=num_epochs, project_name=project_name)
        all_metrics[name] = metrics
    
    # Plot all results
    plot_training_curves(all_metrics)

if __name__ == "__main__":
    main()
```

# ResNet18 Model Analysis and Results

### 1. Comparing Scratch vs Pretrained Models

#### Performance Analysis:

- **Best performing pretrained model:** PretrainedMod2 with **90.32%** test accuracy.
- Clear advantage for pretrained models, with ~**10%** higher accuracy.

#### Training Behavior:
- **Scratch models:**
  - Show steady learning curve.
  - Take longer to converge.
  - Final test accuracies: **76-80%**.
  - Less overfitting (smaller gap between train/test accuracy).

- **Pretrained models:**
  - Faster convergence.
  - Higher final accuracies: **75-90%**.
  - Show more overfitting (larger train/test accuracy gap).
  - **PretrainedMod2** achieves best balance of performance and stability.

### 2. F1 Score and Confusion Matrix Analysis

#### Scratch Models:
- **Mod1:** F1 Score **0.8074** – Shows good diagonal dominance but some class confusion.
- **Mod2:** F1 Score **0.7931** – Similar pattern but more off-diagonal elements.

---

### 3. Initialization Impact Analysis

The different initialization distributions between the new first layer and pretrained layers don't significantly harm model performance, as evidenced by:
- **PretrainedMod2** achieves **90.32%** accuracy despite mixed initialization.
- Training curves show stable learning without convergence issues.
- No signs of gradient problems or learning difficulties.

The success likely comes from:
- **BatchNorm** layers helping normalize activations between differently initialized layers.
- The network's ability to adapt layer interactions during training.
- Pretrained features being generally useful despite initialization differences.

---

### 4. Architecture Impact


-  Kernel Size : Smaller kernels (**3x3**) in initial layers perform better than original **7x7**.

- Multiple Conv Layers: Progressive feature extraction (**PretrainedMod2**) works best.

- Spatial Dimensions: Removing aggressive downsampling helps preserve features.

- Channel Progression: Gradual channel expansion shows better results than immediate expansion to **64** channels.

---

### Key Findings:
1. **Pretraining** provides significant advantage despite mixed initialization.
2. Multiple smaller **convolutions** outperform single large convolution.
3. Careful management of **spatial dimensions** is crucial for small input images.
4. **BatchNorm** plays important role in stabilizing training with different initializations.
5. Best architecture (**PretrainedMod2**) combines domain-specific input processing with pretrained feature extraction.

---

### Differences Arise Primarily From:
1. Better **feature extraction** from pretrained weights.
2. More appropriate architectural choices for the **small input size**.
3. Balanced trade-off between **spatial dimension reduction** and **feature extraction**.
4. Effective handling of the **initialization difference** through proper normalization.
